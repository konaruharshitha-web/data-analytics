```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [2500, 3000, 4200, 5000, 6200, 7000],
    "Profit": [300, 450, 600, 750, 950, 1200],
    "Orders": [120, 135, 160, 185, 220, 250]
}
df = pd.DataFrame(data)
print("Dataset")
print(df)
```

```python
# 1. Line Chart - Sales Trend
plt.figure(figsize=(8,5))
plt.plot(df["Month"], df["Sales"], marker='o', linewidth=2)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
# 2. Bar Chart - Profit Comparison
plt.figure(figsize=(8,5))
plt.bar(df["Month"], df["Profit"])
plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.show()
```

```python
# 3. Scatter Plot
plt.figure(figsize=(8,5))
plt.scatter(df["Sales"], df["Profit"], s=100)
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

```python
# 4. Histogram
plt.figure(figsize=(8,5))
plt.hist(df["Sales"], bins=5)
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
# 5. Pie Chart
plt.figure(figsize=(6,6))
plt.pie(
    df["Sales"],
    labels=df["Month"],
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Sales Contribution by Month")
plt.show()
```

```python
# 6. Seaborn Heatmap
plt.figure(figsize=(6,4))
sns.heatmap(df[["Sales","Profit","Orders"]].corr(),
            annot=True,
            cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
# 7. Seaborn Pairplot
sns.pairplot(df[["Sales","Profit","Orders"]])
plt.show()
```

```python
# DATA STORY
print("\n================ DATA STORY ================\n")
print("1. Sales increase steadily from January to June.")
print("2. Profit also increases with sales.")
print("3. Higher sales generate higher profits.")
print("4. Orders grow consistently every month.")
print("5. Correlation heatmap indicates a strong positive relationship")
print("   between Sales, Profit, and Orders.")
print("6. Business should continue marketing strategies that boosted")
print("   sales in later months.")
print("\n============================================")
```
